# TASK 0

```python
import numpy as np
import csv
```

```python
deliveries = np.genfromtxt(
    "deliveries.csv",
    delimiter=",",
    dtype=None,  
    encoding="utf-8",
    names=True,
    missing_values="",
    filling_values=""
)

with open("matches.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    header = next(reader)  
    data = [row for row in reader]
 
matches = np.array(data, dtype=str)
```

```python
print("Deliveries Columns:", deliveries.dtype.names)
print("Matches Columns:", matches.dtype.names)
```

```python
match_ids = deliveries["match_id"]
batting_team = deliveries["batting_team"]
batter = deliveries["batter"]
bowler = deliveries["bowler"]
batsman_runs = deliveries["batsman_runs"]
overs = deliveries["over"]
```

```python
print("Sample Data:\n")

print("Match IDs:", match_ids[:5])
print("Batting Team:", batting_team[:5])
print("Batter:", batter[:5])
print("Bowler:", bowler[:5])
print("Batsman Runs:", batsman_runs[:5])
print("Overs:", overs[:5])
```

# TASK 1


```python
unique_match_ids = np.unique(match_ids)
```

```python
result = []

for mid in unique_match_ids:
    mask = (match_ids == mid)          
    total = np.sum(batsman_runs[mask]) 
    result.append((mid, total))

result = np.array(result, dtype=object)
```

```python
print(result[:10])
```

# TASK 2


```python
unique_batters, inverse_idx = np.unique(batter, return_inverse=True)
```

```python
total_runs = np.bincount(inverse_idx, weights=batsman_runs)
```

```python
sorted_indices = np.argsort(total_runs)[::-1]
```

```python
top_5_idx = sorted_indices[:5]
top_5_batters = list(zip(unique_batters[top_5_idx], total_runs[top_5_idx]))
```

```python
print(top_5_batters)
```

# TASK 3


```python
unique_batters, inverse_idx = np.unique(batter, return_inverse=True)
```

```python
total_runs = np.bincount(inverse_idx, weights=batsman_runs)
balls_faced = np.bincount(inverse_idx)
strike_rate = (total_runs / balls_faced) * 100
```

```python
result = list(zip(unique_batters, strike_rate))
```

```python
print(result[:10])
```

# TASK 4


```python
unique_bowlers, inverse_idx = np.unique(bowler, return_inverse=True)
```

```python
runs_conceded = np.bincount(inverse_idx, weights=batsman_runs)
balls_bowled = np.bincount(inverse_idx)
overs_bowled = balls_bowled / 6
```

```python
economy = np.divide(
    runs_conceded,
    overs_bowled,
    out=np.zeros_like(runs_conceded, dtype=float),
    where=overs_bowled != 0
)

result = list(zip(unique_bowlers, economy))
```

```python
print(result[:10])
```
